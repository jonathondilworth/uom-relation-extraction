# Training: POS-Tag & Dependency Relation Variations in Entity Markers  

We are testing different **POS-tag and dependency relation variations** within entity markers to improve the model’s **syntactic understanding**.  
This builds upon previous entity marker variations (without POS or dependency tags) explored in  
["An Improved Baseline for Sentence-level Relation Extraction"](https://arxiv.org/abs/1905.08284).  

By incorporating **POS and dependency relation information** within entity markers, we are trying to increase model’s ability to capture **syntactic relationships** more effectively. The experiments will compare model performance across different entity marker variations to assess their impact on relation extraction.



### Example Sentence  
To illustrate the techniques, consider the sentence:  
**"The King of England lives in Buckingham Palace."**  



## New Variations  

### 1. **Typed Entity Marker POS Punct**
– Appends the **first POS tag** using punctuation-based markers.  
   - **Format:**  ```  @ *NER type+first POS tag* subject@ ...# ^NER type + first POS tag^ object# ```  
   - **Example:**  ```  @*PERSON+DT*The King of England@ lives in #^LOCATION+NNP^Buckingham Palace#. ```  

### 2. **Typed Entity Marker DepRel Set Punct**
 – Incorporates **dependency relation (DepRel) tags** for both subject and object entities using a **set of unique dependeny relations**.
   - **Format:**   ```  @ *NER type + DepRel Set* subject@ ... #^NER type + DepRel Set^ object# ```  
   
   - **Example:**   ```  @*PERSON+nsubj set case nmod*The King of England@ lives in #^LOCATION+compound obl^Buckingham Palace#. ```
