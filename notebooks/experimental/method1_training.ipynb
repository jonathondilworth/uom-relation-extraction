# **Training: POS-Tag & Dependency Relation Variations in Entity Markers**  

This model builds upon previous work on entity marker variations explored in  
["An Improved Baseline for Sentence-level Relation Extraction"](https://arxiv.org/abs/1905.08284).  

In this paper, the best performing entity marker variation was the typed entity marker with punctuation.
  - This is the format:  ```  @ *NER type* subject@ ...# ^NER type^ object# ```  
  
We are experimenting with **POS-tag incorporation** within entity markers to enhance the model’s **syntactic understanding**.
By embedding **POS information** into entity markers, we are hoping to improve the model’s ability to capture **syntactic relationships** more effectively.  

---

## **New Variation: Typed Entity Marker POS Punct**  
– Appends the **first POS tag** using punctuation-based markers.  
   - **Format:**
    
      ```  @ *NER type+first POS tag* subject@ ...# ^NER type + first POS tag^ object# ```  
   - **Example:**  
   ```  @*PERSON+DT*The King of England@ lives in #^LOCATION+NNP^Buckingham Palace#. ```
